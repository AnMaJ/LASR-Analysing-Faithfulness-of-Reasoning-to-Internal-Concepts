```python
import gzip
import json

with gzip.open('/workspace/LASR-Analysing-Faithfulness-of-Reasoning-to-Internal-Concepts/index.json.gz', 'rt', encoding='utf-8') as f:
    data = json.load(f)

print(type(data))
print(data.keys())  # if it's a dict

# If nested, see all keys recursively
def print_keys(obj, indent=0):
    if isinstance(obj, dict):
        for k in obj.keys():
            print("  " * indent + str(k))
            print_keys(obj[k], indent + 1)
    elif isinstance(obj, list) and len(obj) > 0:
        print("  " * indent + f"[list of {len(obj)}]")
        print_keys(obj[0], indent + 1)  # peek at first item

print_keys(data)
```
