# Accuracy Metrics Comparison

Load experiment results from `experiment_results/` and generate comparison plots
for CoT vs One-Word and Few-shot vs Zero-shot experiments.

```python
from lasr.plotting import (
    load_experiment_results,
    extract_comparison_metrics,
    plot_experiment_comparison,
)
```

```python
raw = load_experiment_results("experiment_results")
raw
```

```python
metrics = extract_comparison_metrics(raw)
metrics
```

```python
fig = plot_experiment_comparison(metrics)
fig.show()
```
